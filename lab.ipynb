```python
from scraper import fetch_website_contents

import os
from openai import OpenAI
from dotenv import load_dotenv
```

```python
load_dotenv()

openrouter_api_key = os.getenv("OPENROUTER_API_KEY")
openrouter_base_url = os.getenv("OPENROUTER_BASE_URL")

openrouter_client = OpenAI(api_key=openrouter_api_key, base_url=openrouter_base_url)

MODEL = "openai/gpt-6-luna"

system_prompt = """
You are a helpful assistant that analyzes the contents of a LinkedIn profile and its experiences section,
and provides a summary of the person's professional background and skills, in French, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""


def create_user_prompt(profile_url):
    experience_url = f"{profile_url}details/experience/"
    profile_contents = fetch_website_contents(profile_url)
    experience_contents = fetch_website_contents(experience_url)

    return f"""
    Here are the contents of a LinkedIn profile and its experiences section.
    Provide a summary of the person's professional background and skills, in French.
    
    Here is the content of the profile:
    {profile_contents}

    Here is the content of the experiences section:
    {experience_contents}

    """


response = openrouter_client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": system_prompt},
        {
            "role": "user",
            "content": create_user_prompt("https://www.linkedin.com/in/maxime-norval-a6a0ab98/"),
        },
    ],
)

print(response.choices[0].message.content)
```
